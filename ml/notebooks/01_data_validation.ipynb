```python
# ==========================================
# FINAL DATA INTEGRITY CHECK
# ==========================================

# 1. Duplicate disease-symptom pairs
pair_duplicates = disease_symptoms.duplicated(
    subset=["disease_id", "symptom_id"]
).sum()

print("1. Duplicate disease-symptom pairs:", pair_duplicates)


# 2. Matrix symptom columns
matrix_id_columns = ["disease_id", "disease_name"]

matrix_symptom_columns = [
    col for col in matrix.columns
    if col not in matrix_id_columns
]

print("\n2. Number of symptom columns:", len(matrix_symptom_columns))
print("   First 10:", matrix_symptom_columns[:10])


# 3. Matrix values
matrix_values = matrix[matrix_symptom_columns]

unique_values = set(matrix_values.stack().unique())

print("\n3. Unique values in symptom matrix:")
print("  ", sorted(unique_values))


# 4. Disease IDs in matrix
matrix_disease_ids = set(matrix["disease_id"])

missing_from_matrix = disease_ids - matrix_disease_ids
extra_in_matrix = matrix_disease_ids - disease_ids

print("\n4. Disease ID check")
print("   Diseases missing from matrix:", len(missing_from_matrix))
print("   Unknown diseases in matrix:", len(extra_in_matrix))


# 5. Symptom IDs in matrix
matrix_symptom_ids = set(matrix_symptom_columns)

missing_symptoms = symptom_ids - matrix_symptom_ids
extra_symptoms = matrix_symptom_ids - symptom_ids

print("\n5. Symptom ID check")
print("   Symptoms missing from matrix:", len(missing_symptoms))
print("   Unknown symptoms in matrix:", len(extra_symptoms))


# 6. Final summary
print("\n==========================================")
print("FINAL VALIDATION SUMMARY")
print("==========================================")

checks = {
    "Duplicate disease-symptom pairs": pair_duplicates == 0,
    "Matrix has 151 symptoms": len(matrix_symptom_columns) == len(symptom_ids),
    "Matrix values are only 0/1": unique_values <= {0, 1},
    "All diseases exist in matrix": len(missing_from_matrix) == 0 and len(extra_in_matrix) == 0,
    "All symptoms exist in matrix": len(missing_symptoms) == 0 and len(extra_symptoms) == 0,
}

for check, passed in checks.items():
    print(("✓" if passed else "✗"), check)
```
